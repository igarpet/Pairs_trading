```python
import pandas as pd
import numpy as np
import yfinance as yf
```

```python
rf = yf.download("^IRX", start = "2016-01-01", end = "2026-01-01", auto_adjust=False, progress = False)
rf = rf["Close"].squeeze()/100
rf.name="risk_free_rate"
rf.head()
```

```python
rf.to_frame().to_parquet("data/processed/risk_free_rates.parquet", index = True)
```
