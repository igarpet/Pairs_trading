```python
import pandas as pd
import numpy as np
import yfinance as yf
from pathlib import Path
from config import *


```

```python
folders = [
    "data/raw",
    "data/processed",
    "data/cache",
    "figures",
    "results"
]

for folder in folders:
    Path(folder).mkdir(parents=True, exist_ok=True)
```

```python
url = "https://raw.githubusercontent.com/datasets/s-and-p-500-companies/master/data/constituents.csv"

sp500 = pd.read_csv(url)
sp500.head()
```

```python
tickers = sp500["Symbol"].tolist()
tickers = [ticker.replace(".","-") for ticker in tickers]
```

```python
data = yf.download(
    tickers = tickers,
    start = START_DATE,
    end = END_DATE,
    auto_adjust = AUTO_ADJUST,
    progress = True)

```

```python
data.head()
```

```python
prices = data["Close"]
prices.shape
```

```python
report = pd.DataFrame({
    "Missing Values": prices.isna().sum(),
    "Missing %": prices.isna().mean()})

report.sort_values("Missing %", ascending = False)
```

```python
valid = report["Missing %"] < MAX_MISSING_PCT

prices = prices.loc[:, valid]
```

```python
prices.shape
```

```python
prices = prices.ffill()
```

```python
assert prices.isna().sum().sum() == 0
```

```python
prices.to_parquet("data/processed/prices.parquet")
```

```python
print("="*50)
print("MARKET DATA REPORT")
print("="*50)

print(f"Observations : {len(prices):,}")
print(f"Stocks       : {prices.shape[1]}")
print(f"Start        : {prices.index.min().date()}")
print(f"End          : {prices.index.max().date()}")

print("="*50)
```
