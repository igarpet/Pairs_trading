# Module 07 — Walk-Forward Out-of-Sample Backtest

Final 30% out-of-sample walk-forward backtest. Formation parameters and the 40 structurally eligible pairs are frozen from the training sample.


```python
%load_ext autoreload
%autoreload 2

import numpy as np
import pandas as pd

from src.backtest import run_walk_forward_backtest, backtest_summary

pd.set_option("display.max_columns", 100)
```

## 1. Load frozen formation data


```python
train_prices = pd.read_parquet("data/processed/train_prices.parquet")
test_prices = pd.read_parquet("data/processed/test_prices.parquet")
eligible_pairs = pd.read_parquet("data/processed/eligible_pairs.parquet")
cointegrated_pairs = pd.read_parquet("data/processed/cointegrated_pairs.parquet")

assert isinstance(train_prices.index, pd.DatetimeIndex), "train_prices index is not DatetimeIndex."
assert isinstance(test_prices.index, pd.DatetimeIndex), "test_prices index is not DatetimeIndex."
assert train_prices.index.max() < test_prices.index.min(), "Train/test periods overlap."

print("Train:", train_prices.index.min(), "->", train_prices.index.max(), train_prices.shape)
print("Test :", test_prices.index.min(), "->", test_prices.index.max(), test_prices.shape)
print("Eligible pairs:", len(eligible_pairs))
```

## 2. Historical risk-free rate


```python
rf_data = pd.read_parquet("data/processed/risk_free_rates.parquet")

if isinstance(rf_data, pd.DataFrame):
    if rf_data.shape[1] != 1:
        raise ValueError("risk_free_rates.parquet must contain exactly one rate column.")
    risk_free_rates = rf_data.iloc[:, 0]
else:
    risk_free_rates = pd.Series(rf_data)

risk_free_rates.index = pd.to_datetime(risk_free_rates.index)
risk_free_rates = risk_free_rates.sort_index().astype(float)

if risk_free_rates.abs().median() > 1:
    raise ValueError("Risk-free rates appear to be percentages. Divide them by 100 first.")

risk_free_rates.tail()
```

## 3. Final backtest configuration


```python
INITIAL_CAPITAL = 100_000.0
ENTRY_Z = 1.5
TARGET_PROBABILITY = 0.70
MEMORY_WINDOW = 60
MAX_HORIZON_DAYS = 126
N_PATHS = 5000
EWMA_LAMBDA = 0.94
SEED = 42
```

## 4. Run the walk-forward simulation


```python
results = run_walk_forward_backtest(
    train_prices=train_prices,
    test_prices=test_prices,
    eligible_pairs=eligible_pairs,
    cointegrated_pairs=cointegrated_pairs,
    risk_free_rates=risk_free_rates,
    initial_capital=INITIAL_CAPITAL,
    entry_z=ENTRY_Z,
    target_probability=TARGET_PROBABILITY,
    memory_window=MEMORY_WINDOW,
    max_horizon_days=MAX_HORIZON_DAYS,
    n_paths=N_PATHS,
    ewma_lambda=EWMA_LAMBDA,
    seed=SEED,
)

trades = results["trades"]
equity_curve = results["equity_curve"]
skipped_signals = results["skipped_signals"]

print("Completed trades:", len(trades))
print("Skipped records :", len(skipped_signals))
print("Final equity    :", equity_curve["equity"].iloc[-1])
```

## 5. Results and sanity checks


```python
summary = backtest_summary(
    trades=trades,
    equity_curve=equity_curve,
    initial_capital=INITIAL_CAPITAL,
)
summary
```

```python
if not trades.empty:
    display(
        trades[[
            "pair", "entry_date", "exit_date", "entry_z",
            "convergence_horizon_trading_days", "option_calendar_dte",
            "dependent_contracts", "independent_contracts",
            "entry_premium", "exit_value", "pnl",
            "trade_return", "exit_reason",
        ]].head(20)
    )

if not skipped_signals.empty and "reason" in skipped_signals.columns:
    display(skipped_signals["reason"].value_counts())
```

```python
if not trades.empty:
    print("Largest single entry premium:", trades["entry_premium"].max())
    print("Median entry premium:", trades["entry_premium"].median())

print("Maximum concurrent positions:", equity_curve["n_open_positions"].max())
print("Minimum cash balance:", equity_curve["cash"].min())
```

## 6. Save Module 07 outputs


```python
trades.to_parquet("data/processed/trades.parquet", index=False)
equity_curve.to_parquet("data/processed/equity_curve.parquet", index=True)
skipped_signals.to_parquet("data/processed/skipped_signals.parquet", index=False)

print("Saved Module 07 outputs.")
```

```python
# ============================================================
# Module 07 — Post-backtest diagnostics
# Uses existing trades, equity_curve, and skipped_signals only.
# ============================================================


import matplotlib.pyplot as plt


# ------------------------------------------------------------
# 1. Equity curve and maximum drawdown episode
# ------------------------------------------------------------
eq = equity_curve.copy().sort_index()

eq["running_peak"] = eq["equity"].cummax()
eq["drawdown"] = eq["equity"] / eq["running_peak"] - 1.0

trough_date = eq["drawdown"].idxmin()
max_drawdown = float(eq.loc[trough_date, "drawdown"])

peak_date = eq.loc[:trough_date, "equity"].idxmax()
peak_equity = float(eq.loc[peak_date, "equity"])
trough_equity = float(eq.loc[trough_date, "equity"])

recovery = eq.loc[trough_date:]
recovered = recovery[recovery["equity"] >= peak_equity]
recovery_date = recovered.index[0] if not recovered.empty else pd.NaT

print("MAXIMUM DRAWDOWN")
print("----------------")
print(f"Peak date     : {peak_date.date()}")
print(f"Trough date   : {trough_date.date()}")
print(f"Recovery date : {recovery_date.date() if pd.notna(recovery_date) else 'Not recovered by end of test'}")
print(f"Peak equity   : ${peak_equity:,.2f}")
print(f"Trough equity : ${trough_equity:,.2f}")
print(f"Max drawdown  : {max_drawdown:.2%}")


plt.figure(figsize=(12, 5))
plt.plot(eq.index, eq["equity"])
plt.axvline(peak_date, linestyle="--", alpha=0.6)
plt.axvline(trough_date, linestyle="--", alpha=0.6)
plt.title("Out-of-Sample Equity Curve")
plt.xlabel("Date")
plt.ylabel("Portfolio Equity ($)")
plt.grid(alpha=0.25)
plt.show()


plt.figure(figsize=(12, 4))
plt.plot(eq.index, eq["drawdown"])
plt.axvline(trough_date, linestyle="--", alpha=0.6)
plt.title("Portfolio Drawdown")
plt.xlabel("Date")
plt.ylabel("Drawdown")
plt.grid(alpha=0.25)
plt.show()


# ------------------------------------------------------------
# 2. Trades associated with the drawdown period
# ------------------------------------------------------------
if not trades.empty:
    t = trades.copy()

    t["entry_date"] = pd.to_datetime(t["entry_date"])
    t["exit_date"] = pd.to_datetime(t["exit_date"])

    drawdown_trades = t[
        (t["entry_date"] <= trough_date)
        & (t["exit_date"] >= peak_date)
    ].copy()

    drawdown_trades = drawdown_trades.sort_values("pnl")

    print("\nTRADES OVERLAPPING MAX-DRAWDOWN WINDOW")
    print("--------------------------------------")
    print(f"Number of overlapping trades: {len(drawdown_trades)}")
    print(f"Aggregate realized PnL      : ${drawdown_trades['pnl'].sum():,.2f}")

    display(
        drawdown_trades[
            [
                "pair",
                "entry_date",
                "exit_date",
                "entry_z",
                "entry_premium",
                "exit_value",
                "pnl",
                "trade_return",
                "exit_reason",
            ]
        ].head(20)
    )


# ------------------------------------------------------------
# 3. PnL concentration / return asymmetry
# ------------------------------------------------------------
if not trades.empty:
    ranked = trades.sort_values("pnl", ascending=False).copy()

    total_realized_pnl = ranked["pnl"].sum()

    top5_pnl = ranked.head(5)["pnl"].sum()
    top10_pnl = ranked.head(10)["pnl"].sum()

    winners = ranked.loc[ranked["pnl"] > 0, "pnl"].sum()
    losers = ranked.loc[ranked["pnl"] < 0, "pnl"].sum()

    print("\nPNL CONCENTRATION")
    print("-----------------")
    print(f"Total realized PnL : ${total_realized_pnl:,.2f}")
    print(f"Top 5 trade PnL    : ${top5_pnl:,.2f}")
    print(f"Top 10 trade PnL   : ${top10_pnl:,.2f}")

    if total_realized_pnl != 0:
        print(f"Top 5 contribution : {top5_pnl / total_realized_pnl:.2%}")
        print(f"Top 10 contribution: {top10_pnl / total_realized_pnl:.2%}")

    print(f"Gross winning PnL  : ${winners:,.2f}")
    print(f"Gross losing PnL   : ${losers:,.2f}")

    display(
        ranked[
            [
                "pair",
                "entry_date",
                "exit_date",
                "entry_premium",
                "pnl",
                "trade_return",
                "exit_reason",
            ]
        ].head(10)
    )

    print("\nTRADE RETURN DISTRIBUTION")
    print("-------------------------")
    print(trades["trade_return"].describe(
        percentiles=[0.10, 0.25, 0.50, 0.75, 0.90]
    ))


# ------------------------------------------------------------
# 4. Skipped-signal diagnostics
# ------------------------------------------------------------
print("\nSKIPPED SIGNALS")
print("---------------")

if skipped_signals.empty:
    print("No skipped signals.")
elif "reason" in skipped_signals.columns:
    skipped_counts = skipped_signals["reason"].value_counts()
    display(skipped_counts.to_frame("count"))
else:
    print("No 'reason' column found in skipped_signals.")


# ------------------------------------------------------------
# 5. Performance by pair
# ------------------------------------------------------------
if not trades.empty:
    pair_performance = (
        trades.groupby("pair")
        .agg(
            n_trades=("pnl", "size"),
            total_pnl=("pnl", "sum"),
            mean_pnl=("pnl", "mean"),
            mean_trade_return=("trade_return", "mean"),
            median_trade_return=("trade_return", "median"),
            win_rate=("pnl", lambda x: (x > 0).mean()),
        )
        .sort_values("total_pnl", ascending=False)
    )

    print("\nPAIR-LEVEL PERFORMANCE")
    print("----------------------")
    display(pair_performance)

    print("\nTOP 10 PAIRS BY TOTAL PNL")
    display(pair_performance.head(10))

    print("\nBOTTOM 10 PAIRS BY TOTAL PNL")
    display(pair_performance.tail(10))


# ------------------------------------------------------------
# 6. Portfolio utilisation
# ------------------------------------------------------------
print("\nPORTFOLIO UTILISATION")
print("---------------------")
print(f"Maximum concurrent positions : {eq['n_open_positions'].max():.0f}")
print(f"Mean concurrent positions    : {eq['n_open_positions'].mean():.2f}")
print(f"Median concurrent positions  : {eq['n_open_positions'].median():.0f}")
print(f"Minimum cash balance         : ${eq['cash'].min():,.2f}")

if not trades.empty:
    print(f"Largest entry premium        : ${trades['entry_premium'].max():,.2f}")
    print(f"Median entry premium         : ${trades['entry_premium'].median():,.2f}")
```
